## 2. 2D Filamentation Instability Simulation

```python
import numpy as np
import matplotlib.pyplot as plt
from epic.solvers.pic2d import PIC2DSolver

Nx = 100
Ny = 100
Lx = 25.0
Ly = 25.0
N_part = 60000
dt = 0.08
t_end = 20.0

weight = (1.0 * Lx * Ly) / N_part
solver = PIC2DSolver(Nx=Nx, Ny=Ny, Lx=Lx, Ly=Ly, dt=dt)

Nh = N_part // 2
np.random.seed(42)
x1 = np.random.uniform(0.0, Lx, Nh)
y1 = np.random.uniform(0.0, Ly, Nh)
v1 = np.zeros((Nh, 3))
v1[:, 0] = np.random.normal(2.5, 0.4, Nh)

x2 = np.random.uniform(0.0, Lx, Nh)
y2 = np.random.uniform(0.0, Ly, Nh)
v2 = np.zeros((Nh, 3))
v2[:, 0] = np.random.normal(-2.5, 0.4, Nh)

solver.add_species("beam1", q=-weight, m=weight, pos_x=x1, pos_y=y1, vel=v1)
solver.add_species("beam2", q=-weight, m=weight, pos_x=x2, pos_y=y2, vel=v2)
solver.initialize()

print("Simulating 2D Filamentation...")
solver.run(t_end=t_end)

plt.figure(figsize=(8, 7), dpi=110)
plt.imshow(solver.rho, origin="lower", extent=[0, Lx, 0, Ly], cmap="inferno", aspect="auto")
plt.colorbar(label=r"Charge Density $ho(x, y)$")
plt.xlabel(r"x ($c/\omega_{pe}$)")
plt.ylabel(r"y ($c/\omega_{pe}$)")
plt.title(rf"2D Filamentation Density Contours at $t = {t_end}$")
plt.tight_layout()
plt.show()
```
